# Database table previews
This notebook connects to the SQLite DB at `backend/users.db` and runs `SELECT * FROM <table> LIMIT 5` for each table.

```python
import sqlite3
from pathlib import Path

# Try several likely locations for users.db and pick the first that exists
candidates = [
    Path('users.db'),
    Path('backend') / 'users.db',
    Path('../backend') / 'users.db',
    Path.cwd() / 'backend' / 'users.db',
]
db_path = next((p for p in candidates if p.exists()), Path('backend') / 'users.db')
print('DB path:', db_path.resolve())

try:
    conn = sqlite3.connect(str(db_path))
except Exception as e:
    print('Error opening DB:', e)
    raise

conn.row_factory = sqlite3.Row
cur = conn.cursor()
tables = [r[0] for r in cur.execute("SELECT name FROM sqlite_master WHERE type='table' AND name NOT LIKE 'sqlite_%'")]
print('Found tables:', tables)
for t in tables:
    print('\n---', t, '---')
    try:
        rows = cur.execute(f"SELECT * FROM {t} LIMIT 5").fetchall()
        if not rows:
            print('(no rows)')
        else:
            for row in rows:
                print(dict(row))
    except Exception as e:
        print('Error querying', t, e)
conn.close()
```

Notes: If `backend/users.db` does not exist, run the backend to initialize the DB (see `backend/utils/db.py`).
